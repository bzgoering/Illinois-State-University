# Question 1: Dictionaries and Sets

# Part A – Student Score Dictionary

```python
student_scores = {
"Mia": 88,
"Noah": 76,
"Liam": 93
}

student_scores['Emma'] = 90
print(f"Noah's original score: {student_scores['Noah']}")

for student, score in student_scores.items():
    print(f'{student}: {score}')

if student_scores.get('Ava') == None:
    print("Ava: Not found")
else:
    print(student_scores.get('Ava'))
```

# Part B – Set Operations

```python
python_students = {"Mia", "Noah", "Liam", "Emma"}
data_students = {"Liam", "Emma", "Olivia", "Ava"}

intersect_students = python_students & data_students
all_students = python_students | data_students

print(f"Students in both: {intersect_students}")
print(f"All students: {all_students}")
```

# Question 2: Functions with Default and Keyword Arguments 

```python
def calculate_total(price, quantity = 1, discount = 0.0):
    result = price * quantity
    if discount != 0:
        result = result * (1 - discount)
    return result

print(f'Single item: {calculate_total(25.0)}')
print(f'Two items: {calculate_total(25.0, 2)}')
print(f'After discount: {calculate_total(25.0, 2, .20)}')
```

# Question 3: Generator and Exception Handling

# Part A – Generator Function

```python
def multiples_of_five(limit):
    result = 0
    while(result + 5 <= limit):
        result += 5
        yield result
    return result

print(list(multiples_of_five(25)))
```

# Part B – Exception Handling

```python
def convert_to_float(value):
    try:
        result = float (value)
    except (ValueError, TypeError):
        return "Invalid"

    return result

print(convert_to_float('18.5'))
print(convert_to_float('hello'))
print(convert_to_float('(1,2)'))
```

# Question 4: NumPy Array Operations and Indexing

```python
import numpy as np

data = np.array([
[12, 18, 25, 30],
[15, 22, 28, 35],
[10, 20, 24, 32]
])

adjusted = data.copy()
adjusted = adjusted + 5

print(f'Number of dimensions: {data.ndim}')
print(f'Shape: {data.shape}\n')
print(f'Second row: {data[1]}\n')
print(f'First two rows and last two columns:\n{data[:2,-2:]}\n')
print(f'Adjusted array:\n{adjusted}\n')
print(f'Adjusted values at least 30:\n{adjusted[adjusted > 29]}')
```

# Question 5: Random Student Score Analysis with NumPy

```python
import random
import numpy as np

rng = np.random.default_rng(166)
scores = rng.integers(60, 101, size=(4, 2))

student_average = np.mean(scores, axis = 1)
test_average = np.mean(scores, axis = 0)
result = np.where(student_average > 74, 'Pass', 'Review')
high_score = np.max(scores)
low_score = np.min(scores)
student_average_order = np.sort(student_average)

print(f'Scores:\n{scores}')
print(f'\nStudent average: {student_average}')
print(f'Test average: {test_average}\n')
print(f'Results: {result}')
print(f'Highest score: {high_score}')
print(f'Lowest score: {low_score}')
print(f'Sorted averages: {student_average_order}')



```
