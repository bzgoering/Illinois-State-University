### Question 2: Function and Type Conversion

```python
price_text = "12.50"
quantity_text = "3"

#convert string into floats and integers
price = float(price_text)
quantity = int(quantity_text)

#fucntion to calculate the total price
def calculate_total(price, quantity):
    return price*quantity

#gets and prints the output from data
total = calculate_total(price, quantity)
print("Total cost: $%.2f" % total)
```

### Question 3: Classify Student Scores

```python
scores = [88, 72, 95, 61, 84]
count = 0

#gets sum of scores and classifies scores with a comment
for score in scores:
    if score > 89:
        print(f"{score}: Excellent")
        count += 1
    elif score > 79:
        print(f"{score}: Good")
        count += 1
    elif score > 69:
        print(f"{score}: Satisfactory")
        count += 1
    else:
        print(f"{score}: Needs Improvement") 

print(f"Passing scores: {count}")
```

### Question 4: Even Numbers with range() 

```python
even_numbers = []
total = 0

#gets even numbers from [1,21) : 1 inclusive, 21 exlusive
for x in range(2,20):
    if x % 2 == 0:
        total += x
        even_numbers.append(x)

print(f"Even numbers: {even_numbers}")
print(f"Sum: {total}")
```

### Question 5: Lists, Tuples, and Slicing

```python
courses = ["IT 166", "IT 179", "IT 191", "IT 261"]

#prints data from list
print("First Course:", courses[0])
print("Last Course:", courses[-1])
print("First three:", courses[:3])

#edits list by adding and removing a subject
courses.append("IT 276")
courses.remove("IT 179")

print("Updated courses:", courses)
```

### Part B - Tuple Records

```python
student_records = [("Mia", 88), ("Noah", 76), ("Liam", 93)]
total = 0

#prints data from list of tuples and sums up the scores
for name, score in student_records:
    print(f"{name}: {score}")
    total += score

#calculate the average scores and prints
average = total/len(student_records)
print(f"Average score: %.1f" % average)
```
