Problem 1

```python
def multiple_of_three(limit):
    number = 3

    while(number <= limit):
        if number % 3 == 0:
            yield(number)
        number += 1

for number in multiple_of_three(15):
    print(number)
```

Problem 2

```python
def calculate_average(total_text, count_text):
    try:
        total = float(total_text)
        count = int(count_text)
        print(total/count)
    except ValueError:
        print('Error: Enter numeric values')
    except ZeroDivisionError:
        print('Error: Count cannot be zero')

print("test 1")
calculate_average('450', '5')

print("test 2")
calculate_average('hello', '5')

print("test 2")
calculate_average('450', '0')
```

Problem 4

```python
def analyze_scores(filename):
    num_scores = 0
    total = 0
    number = 0

    try:
        with open(filename, mode = 'r', encoding = 'utf-8') as file:
            for line in file:
                number = int(line)
                num_scores+= 1
                total += number

        print("Number of scores: ", num_scores)
        print("Average score: ", total/num_scores)
    except FileNotFoundError:
        print("Error: File not found")

analyze_scores('scores.txt')
analyze_scores('missing.txt')
```
