Question 1: Create and Modify a NumPy Array

```python
import numpy as np

numbers = np.array([10,20,30,40,50])
print(numbers)
print('Shape:',numbers.shape)
print("dtype:", numbers.dtype)
double_numbers = numbers * 2
print(double_numbers)
print(numbers[1:4])
```

Problem 2: Two-Dimensional Indexing and Slicing

```python
import numpy as np

numbers = np.array(
[[10, 20, 30],
[40, 50, 60],
[70, 80, 90]])

print("full array:\n", numbers)
print("second row:\n", numbers[1])
number_eighty = numbers[2,1]
print("Only 80:\n", number_eighty)
print("first 2 rows:\n",numbers[0:2])
print("Transpose:\n", numbers.T)
```

Problem 3: Boolean Indexing and np.where()

```python
import numpy as np

scores = np.array([65, 78, 92, 55, 88, 70])
count_high = np.sum(scores > 69)
results = np.where(scores >= 70, 'Pass', 'Fail')

print(scores[scores > 69])
print("Number passing:", count_high)
print(results)
```

Problem 4: Random Data and Basic Statistics (25 points)

```python
import numpy as np

rng = np.random.default_rng(seed=12345)
data = rng.integers(50, 101, size=(3, 4))
mean = data.mean()
min = data.min()
max = data.max()
row_mean = np.round(np.mean(data, axis = 1), 1)
column_mean = np.mean(data, axis = 0)

print(data)
print(f"Mean: {mean:.2f} Minimum: {min} Maximum: {max}")
print("Row means:", row_mean)
print("Column means:", column_mean)
```
