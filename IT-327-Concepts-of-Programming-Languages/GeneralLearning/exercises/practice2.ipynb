Problem 1

```python

student_tuple = ("Alice", 20, "IT", 92)
courses_tuple = ("IT 166", "IT 191", "IT 166", "IT 261")

print(student_tuple[0], "Major is:", student_tuple[2])

name, num1, major, num2 = student_tuple
print(name, num1, major, num2)

count = courses_tuple.count("IT 166")
print("Count of IT 166:", count)

index = courses_tuple.index("IT 166")
print("Index of first IT 166:", index)
```

Problem 2

```python
courses = ["IT 166", "IT 191", "IT 178"]

courses[1] = "IT 168"
courses.append("IT 261")
courses.insert(1, "IT 179")
courses.remove("IT 178")
removed = courses.pop()

count = courses.count("IT 166")
if count == 0:
    print("IT 166 was removed")
else:
    print("IT 166 was not removed")

print("Removed:", removed)
print(courses)
```

Problem 3

```python
scores = [88,92,76,95,84,90]

print(scores[:3])
print(scores[-3:])
print(scores[::2])
print(scores[::-1])
updated_scores = scores[:]
updated_scores[2] = 80
updated_scores[3] = 96

print(scores)
print(updated_scores)
```

Problem 4

```python
records = [("Alice", 92), ("Bob", 78), ("Carol", 85), ("David", 69)]

print(records[0])

name,score = records[0]
print(name,score)

count_score = 0
high_scores = []
for name,score in records:
    count_score += score
    print(name, score)
    if score > 79:
        high_scores.append(name)

print("average = ", count_score/4, "; high_scores = ", high_scores, sep = "")



```
